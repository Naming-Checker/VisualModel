```python
import os
import pandas as pd
```

```python
def fix_file(path, to_path):
    frame = pd.read_csv(path, sep=' ', header=None, index_col=None)
    frame = frame.transpose().map(lambda p: os.path.basename(p))
    frame.to_csv(to_path, header=None, index=False)
    return frame
```

```python
fix_file('../data/logos.csv', '../data/logos_post.csv')
```

```python
fix_file('../data/text_only.csv', '../data/text_post.csv')
```
