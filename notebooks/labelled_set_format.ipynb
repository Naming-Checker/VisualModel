```python
import os
import pandas as pd
```

```python
def fix_file(path, to_path_base):
    frame = pd.read_csv(path, sep=' ', header=None, index_col=None)
    frame = frame.transpose().map(lambda p: os.path.basename(p))
    frame.to_csv(f'{to_path_base}_all.csv', header=None, index=False)
    train_size = (int)(len(frame) * 0.8)
    frame.head(train_size).to_csv(f'{to_path_base}_train.csv', header=None, index=False)
    frame.tail(len(frame) - train_size).to_csv(f'{to_path_base}_valid.csv', header=None, index=False)
    return frame
```

```python
fix_file('../data/logos.csv', '../data/logos')
```

```python
fix_file('../data/text_only.csv', '../data/text')
```
